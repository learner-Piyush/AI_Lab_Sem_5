## 1. Write a python program to implement best first search algorithm using greedy approach

```python
import heapq

def best_first_search(graph, heuristic, start, goal):
    priority_queue = []
    heapq.heappush(priority_queue, (heuristic[start], start))

    visited = set()

    while priority_queue:
        h, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        print(current, end=" ")
        visited.add(current)

        if current == goal:
            print("\nGoal Found!")
            return

        for neighbour in graph[current]:
            if neighbour not in visited:
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbour], neighbour)
                )

    print("\nGoal Not Found!")


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

heuristic = {
    'A': 10,
    'B': 8,
    'C': 5,
    'D': 7,
    'E': 3,
    'F': 2,
    'G': 0
}

start = 'A'
goal = 'G'

print("Best First Search Traversal:")
best_first_search(graph, heuristic, start, goal)
```

## 2. Write a python program to implement A* algorithm

```python
import heapq

def a_star(graph, heuristic, start, goal):
    priority_queue = []
    heapq.heappush(priority_queue, (heuristic[start], 0, start, [start]))

    visited = set()

    while priority_queue:
        f, g, current, path = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            print("Shortest Path:", " -> ".join(path))
            print("Total Cost:", g)
            return

        for neighbour, cost in graph[current]:
            if neighbour not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbour]

                heapq.heappush(
                    priority_queue,
                    (new_f, new_g, neighbour, path + [neighbour])
                )

    print("Goal Not Found!")


graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 4)],
    'E': [('G', 2)],
    'F': [('G', 5)],
    'G': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 5,
    'D': 4,
    'E': 2,
    'F': 4,
    'G': 0
}

start = 'A'
goal = 'G'

print("A* Search Algorithm:")
a_star(graph, heuristic, start, goal)
```

## 3. Write a python program for hill climbing algorithm

```python
def hill_climbing(graph, heuristic, start, goal):
    current = start

    print("Hill Climbing Path:")

    while True:
        print(current, end=" ")

        if current == goal:
            print("\nGoal Reached!")
            return

        neighbours = graph[current]

        if not neighbours:
            print("\nNo Better Neighbour Found!")
            return

        next_node = min(neighbours, key=lambda x: heuristic[x])

        if heuristic[next_node] >= heuristic[current]:
            print("\nReached Local Optimum!")
            return

        current = next_node


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

heuristic = {
    'A': 10,
    'B': 8,
    'C': 5,
    'D': 7,
    'E': 3,
    'F': 2,
    'G': 0
}

start = 'A'
goal = 'G'

hill_climbing(graph, heuristic, start, goal)
```
