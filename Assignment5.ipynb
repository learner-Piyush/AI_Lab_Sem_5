## 1. Write a python program to get the levels in MultiIndex.

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('India', 'West Bengal'),
     ('India', 'Maharashtra'),
     ('USA', 'California'),
     ('USA', 'Texas')]
)

print("MultiIndex:")
print(index)

print("\nLevels:")
print(index.levels)
```

## 3. Write a python program to get the name of levels in MultiIndex.

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('India', 'West Bengal'),
     ('India', 'Maharashtra'),
     ('USA', 'California'),
     ('USA', 'Texas')],
    names=['Country', 'State']
)

print("Names of MultiIndex levels:")
print(index.names)
```

## 4. Write a python program to Set levels on a MultiIndex.

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('India', 'WB'),
     ('India', 'MH'),
     ('USA', 'CA'),
     ('USA', 'TX')],
    names=['Country', 'State']
)

print("Original MultiIndex:")
print(index)

new_index = index.set_levels(
    [['India', 'USA'], ['West Bengal', 'Maharashtra', 'California', 'Texas']]
)

print("\nAfter Setting Levels:")
print(new_index)
```

## 5. Write a python program to Swap levels of a MultiIndex.

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('India', 'West Bengal'),
     ('India', 'Maharashtra'),
     ('USA', 'California'),
     ('USA', 'Texas')],
    names=['Country', 'State']
)

print("Original MultiIndex:")
print(index)

swapped = index.swaplevel()

print("\nAfter Swapping Levels:")
print(swapped)
```

## 6. Write a python program to Rearrange levels using level name in MultiIndex.

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('India', 'West Bengal', 'Kolkata'),
     ('India', 'Maharashtra', 'Mumbai'),
     ('USA', 'California', 'Los Angeles'),
     ('USA', 'Texas', 'Houston')],
    names=['Country', 'State', 'City']
)

print("Original MultiIndex:")
print(index)

rearranged = index.reorder_levels(['City', 'Country', 'State'])

print("\nAfter Rearranging Levels:")
print(rearranged)
```

## 7. Write a python program to Adding new column to existing Data-Frame by using Data-Frame. `insert()`.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit'],
    'Marks': [85, 78, 92]
})

print("Original DataFrame:")
print(df)

df.insert(1, 'Age', [20, 21, 20])

print("\nAfter Adding New Column:")
print(df)
```

## 8. Write a python program to Adding new column to existing Data-Frame by using a dictionary.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit'],
    'Marks': [85, 78, 92]
})

print("Original DataFrame:")
print(df)

new_data = {
    'Age': [20, 21, 20]
}

df = pd.DataFrame({**df.to_dict('series'), **new_data})

print("\nAfter Adding New Column:")
print(df)
```
