## 1. Write a python program to create NumPy array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("NumPy Array:")
print(arr)
```

## 2. Write a python program to add a list to NumPy array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
lst = [1, 2, 3, 4, 5]

result = arr + lst

print("NumPy Array:", arr)
print("List:", lst)
print("After Addition:", result)
```

## 3. Write a python program to delete values from a NumPy array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original Array:")
print(arr)

arr = np.delete(arr, [1, 3])

print("\nAfter Deleting Values at Index 1 and 3:")
print(arr)
```

## 4. Write a python program to calculate the sum of every row in a NumPy array.

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Array:")
print(arr)

row_sum = np.sum(arr, axis=1)

print("\nSum of Every Row:")
print(row_sum)
```

## 5. Write a python program to check if a value exists in NumPy Array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

value = int(input("Enter value to search: "))

if value in arr:
    print(value, "exists in the NumPy array.")
else:
    print(value, "does not exist in the NumPy array.")
```

## 6. Write a python program to initialize a NumPy array.

```python
import numpy as np

# Initialize an array with zeros
zeros = np.zeros(5)

# Initialize an array with ones
ones = np.ones(5)

# Initialize an array with a range of values
numbers = np.arange(1, 6)

print("Array of Zeros:")
print(zeros)

print("\nArray of Ones:")
print(ones)

print("\nArray using arange():")
print(numbers)
```

## 7. Write a python program to create a simple line chart using the `plot()` function.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.title("Simple Line Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

## 8. Write a python program to create a simple pie chart using the `pie()` function.

```python
import matplotlib.pyplot as plt

labels = ['Python', 'Java', 'C++', 'JavaScript']
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct='%1.1f%%')

plt.title("Programming Language Usage")

plt.show()
```

## 9. Write a python program to create a simple Violin plot.

```python
import matplotlib.pyplot as plt

marks = [55, 60, 65, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.violinplot(marks)

plt.title("Simple Violin Plot")
plt.ylabel("Marks")

plt.show()
```

## 10. Write a python program to create a Scatter plot.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [10, 15, 12, 20, 18, 25, 22, 30]

plt.scatter(x, y)

plt.title("Simple Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

## 11. Write a python program to display a histogram.

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75,
         78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(marks, bins=5)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```
