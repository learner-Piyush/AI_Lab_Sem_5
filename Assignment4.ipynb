## 1. Write a Python program to create a function to accept a list of numbers and return unique element (displaying 1 time).

```python
def unique_elements(numbers):
    result = []

    for num in numbers:
        if numbers.count(num) == 1:
            result.append(num)

    return result

numbers = [10, 20, 30, 20, 40, 50, 30, 60]

print("Original List:", numbers)
print("Unique Elements:", unique_elements(numbers))
```

## 2. Write a Python program to create different function to implement choice based calculator. Functions shall be created for addition, subtraction, multiplication, division operation.

```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    return a / b

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("\n1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your choice: "))

if choice == 1:
    print("Result:", addition(a, b))
elif choice == 2:
    print("Result:", subtraction(a, b))
elif choice == 3:
    print("Result:", multiplication(a, b))
elif choice == 4:
    if b != 0:
        print("Result:", division(a, b))
    else:
        print("Division by zero is not possible.")
else:
    print("Invalid choice.")
```

## 3. Write a Python program to create a recursive function to compute factorial of a number.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

number = int(input("Enter a number: "))

print("Factorial:", factorial(number))
```

## 4. Write a Python program to create a recursive function to compute particular Fibonacci series term and return it. Range of series terms shall be passed to the created function.

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

start = int(input("Enter starting term: "))
end = int(input("Enter ending term: "))

print("Fibonacci Series:")

for i in range(start, end + 1):
    print(fibonacci(i), end=" ")
```

## 5. Write a Python program to create a function to accept dictionary of student information and show call by reference implementation.

```python
def update_student(student):
    student["marks"] = 95
    student["grade"] = "A"


student = {
    "name": "Piyush",
    "roll": 72,
    "marks": 85
}

print("Before Function Call:")
print(student)

update_student(student)

print("\nAfter Function Call:")
print(student)
```

## 7. Write a Python program to create a generator function to display Fibonacci series terms between 1 and 100

```python
def fibonacci_generator():
    a, b = 0, 1

    while a <= 100:
        if a >= 1:
            yield a
        a, b = b, a + b

print("Fibonacci Series between 1 and 100:")

for term in fibonacci_generator():
    print(term, end=" ")
```

## 8. Write a Python program to use lambda and filter function to print even numbers among a list of numbers.

```python
numbers = [10, 15, 22, 31, 40, 47, 56, 63, 70]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original List:", numbers)
print("Even Numbers:", even_numbers)
```
