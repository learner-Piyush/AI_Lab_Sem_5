## 1. Write a python program to take element into a list and show the positive index and negative index.

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter element: "))
    numbers.append(num)

print("\nElement\tPositive Index\tNegative Index")

for i in range(len(numbers)):
    print(numbers[i], "\t", i, "\t\t", i - len(numbers))
```

## 3. Write a python program to sort the element into a list in increasing order and decreasing order.

```python
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter element: "))
    numbers.append(num)

ascending = sorted(numbers)
descending = sorted(numbers, reverse=True)

print("Ascending Order:", ascending)
print("Descending Order:", descending)
```

## 4. Write a python program to find element which repeated in a list with their index number.

```python
numbers = [10, 20, 30, 20, 40, 30, 50, 20]

checked = []

for value in numbers:
    if value not in checked and numbers.count(value) > 1:
        print(value, "repeated at indices:", end=" ")
        for i in range(len(numbers)):
            if numbers[i] == value:
                print(i, end=" ")
        print()
        checked.append(value)
```

## 7. Write a python program to print even numbers from a list of elements.

## 8. Write a python program to print prime numbers from a list.

```python
numbers = [2, 5, 8, 11, 15, 17, 20, 23]

print("Prime Numbers:")

for num in numbers:
    if num > 1:
        prime = True

        for i in range(2, num):
            if num % i == 0:
                prime = False
                break

        if prime:
            print(num)
```

## 9. Write a python program to check whether a number is perfect number or not.

```python
number = int(input("Enter a number: "))

sum = 0

for i in range(1, number):
    if number % i == 0:
        sum += i

if sum == number:
    print(number, "is a Perfect Number.")
else:
    print(number, "is NOT a Perfect Number.")
```

## 10. Write a python program to combined multiple lists.

EX:
- Person= [‘i’, ’you]
- Choice= [‘like’, ’hate]
- Hobby= [‘swimming’, ‘driving’]

Output:
- I like swimming
- I like driving
- I hate swimming
- I hate driving
- you like swimming
- you like driving
- you hate swimming
- you hate driving

```python
Person = ["I", "You"]
Choice = ["like", "hate"]
Hobby = ["swimming", "driving"]

for person in Person:
    for choice in Choice:
        for hobby in Hobby:
            print(person, choice, hobby)
```
