## 1. Write a python program for adding, removing and modifying list item using slicing operation. And for inserting multiple items at 3 indexes, for deleting some element from list with proper index value.

```python
# Initial List
my_list = [10, 20, 30, 40, 50, 60]

print("Original List:", my_list)

# Adding multiple items using slicing
my_list[2:2] = [25, 27]
print("\nAfter Adding:", my_list)

# Modifying items using slicing
my_list[1:4] = [200, 300, 400]
print("After Modifying:", my_list)

# Removing items using slicing
my_list[3:5] = []
print("After Removing:", my_list)

# Inserting multiple items at index 3
my_list[3:3] = [500, 600, 700]
print("After Inserting at Index 3:", my_list)

# Deleting an element using index
del my_list[2]
print("After Deleting Element at Index 2:", my_list)
```

## 2. Write a python program to sort multiple strings and adding the element to the list using `for` loop. Sorting and displaying the new list outside the `for` loop.

```python
# Empty list
words = []

# Predefined strings
data = ["Banana", "Apple", "Orange", "Mango", "Grapes"]

# Adding elements using for loop
for word in data:
    words.append(word)

# Sorting outside the loop
words.sort()

print("Sorted List:")
print(words)
```

## 3. Write a python program to sort dictionary item by values with that sorting dictionary item in ascending order and descending order and displaying the result.

```python
student_marks = {
    "Rahul": 78,
    "Amit": 90,
    "Priya": 85,
    "Neha": 65,
    "Riya": 95
}

# Ascending Order
ascending = dict(sorted(student_marks.items(), key=lambda item: item[1]))

# Descending Order
descending = dict(sorted(student_marks.items(), key=lambda item: item[1], reverse=True))

print("Ascending Order:")
print(ascending)

print("\nDescending Order:")
print(descending)
```

## 5. Write a python program to print index at which a particular value exists. If the value exists a multiple location in the list, then print all the indices. Also count the number of times that value is repeated in the list.

```python
numbers = [10, 20, 30, 20, 40, 20, 50, 60, 20]

value = int(input("Enter value to search: "))

count = 0

print("Indices:")

for index in range(len(numbers)):
    if numbers[index] == value:
        print(index)
        count += 1

print("Total Occurrences:", count)
```

## 6. Write a python program that creates two sets-Squares and Cubes in range 1-10. Demonstrate the using of `update()`, `pop()`, `remove()`, `add()` and `clear()` functions.

```python
# Squares
Squares = {x**2 for x in range(1, 11)}

# Cubes
Cubes = {x**3 for x in range(1, 11)}

print("Squares:", Squares)
print("Cubes:", Cubes)

# update()
Squares.update(Cubes)
print("\nAfter update():")
print(Squares)

# add()
Squares.add(999)
print("\nAfter add():")
print(Squares)

# pop()
removed = Squares.pop()
print("\nPopped Element:", removed)
print(Squares)

# remove()
Squares.remove(999)
print("\nAfter remove(999):")
print(Squares)

# clear()
Cubes.clear()
print("\nAfter clear():")
print(Cubes)
```

## 7. Write a python program using `filter` function to a list of squares of number from 1-10. Then using the `for in` construct to sum the elements on the list generated.

```python
# Numbers from 1 to 10
numbers = list(range(1, 11))

# Square each number
squares = list(map(lambda x: x*x, numbers))

# Filter even squares
filtered = list(filter(lambda x: x % 2 == 0, squares))

print("Filtered Squares:")
print(filtered)

# Sum using for loop
total = 0

for value in filtered:
    total += value

print("Sum =", total)
```
