## 1. write a python program to Arithmetic Operations with NumPy Array.

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 5, 8])

print("Array 1:", arr1)
print("Array 2:", arr2)

print("\nAddition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)
print("Division:", arr1 / arr2)
```

## 2. write a python program to sum all elements in a NumPy array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("NumPy Array:", arr)
print("Sum of all elements:", np.sum(arr))
```

## 3. write a python program to convert 1D array to 2D array.

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

print("1D Array:")
print(arr)

arr_2d = arr.reshape(2, 3)

print("\n2D Array:")
print(arr_2d)
```

## 4. write a python program to convert List or Tuple into NumPy array.

```python
import numpy as np

my_list = [10, 20, 30, 40, 50]
my_tuple = (60, 70, 80, 90, 100)

array_from_list = np.array(my_list)
array_from_tuple = np.array(my_tuple)

print("List:", my_list)
print("NumPy Array from List:", array_from_list)

print("\nTuple:", my_tuple)
print("NumPy Array from Tuple:", array_from_tuple)
```

## 5. write a python program to append a NumPy array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print("Original Array:", arr)

arr = np.append(arr, [50, 60])

print("After Appending:", arr)
```

## 6. write a python program to sort a NumPy array in descending order.

```python
import numpy as np

arr = np.array([50, 10, 40, 30, 20])

print("Original Array:", arr)

ascending = np.sort(arr)
descending = ascending[::-1]

print("Ascending Order:", ascending)
print("Descending Order:", descending)
```

## 7. Write a python program to create a bar chart.

```python
import matplotlib.pyplot as plt

subjects = ['Python', 'Java', 'C++', 'DBMS']
marks = [85, 78, 90, 82]

plt.bar(subjects, marks)

plt.title("Subject-wise Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

## 8. Write a python program to create a quiver plot.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3])
y = np.array([0, 0, 0, 0])

u = np.array([1, 1, 1, 1])
v = np.array([1, 2, 3, 4])

plt.quiver(x, y, u, v)

plt.title("Quiver Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

## 9. Write a python program to create a meshgrid chart.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 20)
y = np.linspace(-5, 5, 20)

X, Y = np.meshgrid(x, y)

plt.scatter(X, Y, s=10)

plt.title("Meshgrid Chart")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

## 10. Write a python program to create a contour plot.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

plt.contour(X, Y, Z)

plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

## 11. Write a python program to create an area plot.

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [20, 35, 30, 45, 50, 60]

plt.fill_between(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
