## 1. Write a python program to Delete a column from a Pandas DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit'],
    'Age': [20, 21, 20],
    'Marks': [85, 78, 92]
})

print("Original DataFrame:")
print(df)

df = df.drop('Age', axis=1)

print("\nAfter Deleting Age Column:")
print(df)
```

## 2. Write a python program to rename columns in Pandas DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit'],
    'Age': [20, 21, 20],
    'Marks': [85, 78, 92]
})

print("Original DataFrame:")
print(df)

df = df.rename(columns={
    'Name': 'Student_Name',
    'Marks': 'Student_Marks'
})

print("\nAfter Renaming Columns:")
print(df)
```

## 3. Write a python program to change the order of DataFrame columns.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit'],
    'Age': [20, 21, 20],
    'Marks': [85, 78, 92]
})

print("Original DataFrame:")
print(df)

df = df[['Marks', 'Name', 'Age']]

print("\nAfter Changing Column Order:")
print(df)
```

## 4. Write a python program to Creating an empty Pandas DataFrame, then filling it.

```python
import pandas as pd

df = pd.DataFrame()

print("Empty DataFrame:")
print(df)

df['Name'] = ['Piyush', 'Rahul', 'Amit']
df['Age'] = [20, 21, 20]
df['Marks'] = [85, 78, 92]

print("\nFilled DataFrame:")
print(df)
```

## 5. Write a python program to convert pandas DataFrame to NumPy array.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit'],
    'Marks': [85, 78, 92]
})

print("DataFrame:")
print(df)

array = df.to_numpy()

print("\nNumPy Array:")
print(array)
```

## 6. Write a python program to merge two DataFrames by index.

```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Piyush', 'Rahul', 'Amit']
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    'Marks': [85, 78, 92]
}, index=[1, 2, 3])

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

merged = pd.merge(df1, df2, left_index=True, right_index=True)

print("\nMerged DataFrame:")
print(merged)
```

## 7. Write a python program to insert rows in pandas DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Piyush', 'Rahul'],
    'Marks': [85, 78]
})

print("Original DataFrame:")
print(df)

df.loc[len(df)] = ['Amit', 92]

print("\nAfter Inserting Row:")
print(df)
```

## 8. Write a python program to replace a character in all column names in Pandas.

```python
import pandas as pd

df = pd.DataFrame({
    'Student_Name': ['Piyush', 'Rahul', 'Amit'],
    'Student_Age': [20, 21, 20],
    'Student_Marks': [85, 78, 92]
})

print("Original Column Names:")
print(df.columns)

df.columns = df.columns.str.replace('_', ' ')

print("\nNew Column Names:")
print(df.columns)

print("\nDataFrame:")
print(df)
```
